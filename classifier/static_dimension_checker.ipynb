```python
REF_L = 2.3734
REF_W = 2.8706
REF_H = 3.5695

VISUAL_THRESHOLD = 0.20
GEOMETRIC_THRESHOLD = 0.20


def check_dimensions(name: str, X: float, Y: float, Z: float) -> None:
    refs = {"X": (X, REF_L), "Y": (Y, REF_W), "Z": (Z, REF_H)}
    deltas = {}
    print(f"\n{'='*60}")
    print(f"{name}  (X={X:.4f}, Y={Y:.4f}, Z={Z:.4f})")
    print(f"Reference: trafficwarning  (L={REF_L}, W={REF_W}, H={REF_H})")
    print(f"{'-'*60}")

    for dim, (val, ref) in refs.items():
        delta = abs(val - ref) / ref
        deltas[dim] = delta
        lo = ref * (1 - VISUAL_THRESHOLD)
        hi = ref * (1 + VISUAL_THRESHOLD)

        if delta <= VISUAL_THRESHOLD:
            headroom = (VISUAL_THRESHOLD - delta) * 100
            status = f"OK, {headroom:.1f}pp headroom"
        else:
            overshoot = (delta - VISUAL_THRESHOLD) * 100
            status = f"OVER by {overshoot:.1f}pp"

        print(f"  delta_{dim} = {delta*100:5.1f}%  [{status}]   valid range: {lo:.4f} .. {hi:.4f}")

    worst_dim = max(deltas, key=deltas.get)
    delta_max = deltas[worst_dim]

    if delta_max <= VISUAL_THRESHOLD:
        verdict = "VISUAL SHIFT (Level 1)"
    elif delta_max > GEOMETRIC_THRESHOLD:
        verdict = "GEOMETRIC SHIFT (Level 2)"
    else:
        verdict = "AMBIGUOUS (unreachable: thresholds meet at 20%)"

    print(f"{'-'*60}")
    print(f"  delta_max = {delta_max*100:.1f}% ({worst_dim}) -> {verdict}")
    print(f"{'='*60}")
```
