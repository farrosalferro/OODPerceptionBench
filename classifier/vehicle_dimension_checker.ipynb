```python
# vehicle_car cluster stats (N=25, from distribution_analysis.ipynb)
MU_L, SIGMA_L = 4.529180, 0.722124
MU_W, SIGMA_W = 1.951264, 0.163473
MU_H, SIGMA_H = 1.573888, 0.170715

Z_VISUAL = 2.0
Z_GEOMETRIC = 3.0
HEIGHT_SATURATION = 2.0


def check_vehicle(name: str, L: float, W: float, H: float) -> None:
    stats = {"L": (L, MU_L, SIGMA_L), "W": (W, MU_W, SIGMA_W), "H": (H, MU_H, SIGMA_H)}
    z_scores = {}

    print(f"\n{'='*64}")
    print(f"{name}  (L={L:.4f}, W={W:.4f}, H={H:.4f})")
    print(f"Reference cluster: vehicle_car  (mu_L={MU_L:.4f}, mu_W={MU_W:.4f}, mu_H={MU_H:.4f})")
    print(f"{'-'*64}")

    for dim, (val, mu, sigma) in stats.items():
        saturated = False
        if dim == "H" and mu > HEIGHT_SATURATION and val > HEIGHT_SATURATION:
            z = 0.0
            saturated = True
        else:
            z = abs(val - mu) / sigma

        z_scores[dim] = z
        lo = mu - Z_VISUAL * sigma
        hi = mu + Z_VISUAL * sigma

        if saturated:
            status = "SATURATED (both > 2.0m, Z forced to 0)"
        elif z <= Z_VISUAL:
            headroom = Z_VISUAL - z
            status = f"OK, headroom {headroom:.2f}σ"
        else:
            overshoot = z - Z_VISUAL
            status = f"OVER by {overshoot:.2f}σ"

        print(f"  Z_{dim} = {z:5.2f}  [{status}]   visual range: {lo:.4f} .. {hi:.4f}")

    worst_dim = max(z_scores, key=z_scores.get)
    z_max = z_scores[worst_dim]

    if z_max <= Z_VISUAL:
        verdict = "VISUAL SHIFT (Level 1)"
    elif z_max > Z_GEOMETRIC:
        verdict = "GEOMETRIC SHIFT (Level 2)"
    else:
        verdict = "AMBIGUOUS"

    print(f"{'-'*64}")
    print(f"  Z_max = {z_max:.2f} ({worst_dim}) -> {verdict}")
    print(f"{'='*64}")
```
